## 2- Importation des données infractus du myocarde

Ce dataset contient les données de 1700 patients dont chacun a eu un infractus. Il compte 124 colonnes 

```python
df_infarctus = pd.read_csv("Myocardial_infarction_complications_Database.csv", na_values="?",low_memory=False)
df_infarctus = df_infarctus.set_index("ID")

print(df_infarctus.shape)   # (1700, 124)
df_infarctus.head()
```

```python
# Sélection des colonnes pertinentes pour l'analyse
columns_to_keep = [
    "AGE", "SEX", "HYPERTENSION", "DIABETES", "SMOKING", "OBESITY", "HYPERLIPIDEMIA", "PREVIOUS_INFARCTION", "STROKE", "ATRIAL_FIBRILLATION",
    "KILLIP_CLASS", "LVEF", "TROPONIN_LEVEL"]

X_df = df_infarctus.iloc[:, 2:113] # colonnes pour la prédiction
y = df_infarctus.iloc[:, 113:125].values # colonnes des complications (outputs)

# data de prédictions pour t=juste après admission
cols_to_drop = [93, 94, 95, 100, 101, 102, 103, 104, 105]
X_admission_df = df_infarctus.drop(df_infarctus.columns[cols_to_drop], axis=1).iloc[:, 2:104]
X_admission = X_admission_df.values

X_df = X_df.values
```

```python
# ACP

inf_pca = PCA()
inf_pca.fit(X_df)
inf_pca.explained_variance_ratio_
plt.bar(range(X_df.shape[1]), inf_pca.explained_variance_ratio_, color="r", align="center")
```
